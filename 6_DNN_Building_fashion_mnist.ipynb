## Stage 3: Dataset preprocessing



### Loading the dataset

```python
# (X_train, y_train), (X_test, y_test) 
(X_train, y_train), (X_test, y_test) = fashion_mnist.load_data()
```

### Image normalization



```python
X_train.max()
```

```python
X_train.min()
```

```python
# 정규화 0~1 사이값으로 정규화
X_train = X_train/255.0   # float으로
```

```python
# 정규화 0~1 사이값으로 정규화
X_test = X_test/255.0
```

### Reshaping of the dataset



```python
28*28
```

```python
X_train = X_train.reshape(60000, 28*28)
X_train.shape
```

```python
X_test = X_test.reshape(10000, 28*28)
X_test.shape
```

## Stage 4: Building an Artificial Neural network

### Defining the model



```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, Flatten
```

### Adding the first layer (Dense layer)

Layer hyper-parameters:
- number of units/neurons: 128
- activation function: ReLU
- input_shape: (784, )

### Adding a Dropout layer 



### Adding the second layer (output layer)

- units == number of classes (10 in the case of Fashion MNIST)
- activation = 'softmax'

### Comiling the model

- Optimizer: Adam
- Loss: Sparse softmax (categorical) crossentropy 

```python
def build_model(input_dim):
    model = Sequential()
    model.add(keras.Input(shape=(input_dim,)))
    model.add(Dense(128, activation=keras.activations.relu ))
    model.add(Dropout(0.2))   # 오버피팅을 감소시키기 위해서. 
    model.add(Dense(10, activation=keras.activations.softmax ))
    model.compile(optimizer=keras.optimizers.Adam(), 
                  loss=keras.losses.SparseCategoricalCrossentropy(),  # label을 one-hot-encoding을 안해도 됨
                  metrics=[keras.metrics.SparseCategoricalAccuracy()])
    return model
```

```python
tf.__version__
```

### Training the model

```python
input_dim = 784
model = build_model(input_dim=input_dim)
```

```python
# validation_data=(X_test, y_test) : 학습하면서 쪽지시험 데이터를 주는 것
# history = model.fit(X_train, y_train, epochs=100, validation_data=(X_test, y_test))
history = model.fit(X_train,
                    y_train,
                    batch_size=1000,
                    epochs=100,
                    validation_split=0.2
)
```

## Stage 6 : 일반화 

```python
# 이미지
t_shirt_image = "images/t-shirt.jpg"
# t_shirt_image = "images/t-shirt2.jpg"
```

```python
from PIL import Image
import numpy as np

import matplotlib.pyplot as plt

# Fashion MNIST 클래스 이름
class_names = ["T-shirt/top", "Trouser", "Pullover", "Dress", "Coat",
         "Sandal", "Shirt", "Sneaker", "Bag", "Ankle boot"]

# 이미지 로드
img = Image.open(t_shirt_image).convert("L")  # 흑백 변환
img = img.resize((28, 28))  # 모델 입력 크기에 맞춤

# Fashion MNIST와 반대 명암의 이미지라면 반전
# 검은 배경 + 흰 피사체 -> 흰 배경 + 검은 피사체
img = img.point(lambda p: 255 - p)

# 모델 입력용 정규화
img_array = np.asarray(img, dtype=np.float32) / 255.0
img_input = img_array.reshape(1, 28 * 28)

# 반전된 이미지 확인
plt.imshow(img, cmap='gray')
plt.show()
```

```python
# 예측
pred = model.predict(img_input, verbose=0)
pred_idx = np.argmax(pred[0])
print("pred", pred)
print("pred_idx", pred_idx)
```

```python
pred_label = class_names[pred_idx]
pred_prob = pred[0, pred_idx]

print(f"예측 클래스: {pred_label}")
print(f"확률: {pred_prob:.2%}")

# 이미지 시각화
plt.figure(figsize=(2, 2))
plt.imshow(img_array, cmap="gray")
plt.title(f"Predicted: {pred_label}\nProb: {pred_prob:.2%}")
plt.axis("off")
plt.show()
```
